```python
#Charger les données du fichier Json dans une variable de type dictionnaire.
import json
with open('intents_chatbot_orientation.json', 'r', encoding='UTF-8') as file:
    data = json.load(file)
print(json.dumps(data, indent=2, ensure_ascii=False))
```

```python
# Afficher toutes les intentions dans le dictionnaire.

for intent in data["intents"]:
  print(intent['tag'])
```

```python
#Une List qui contient toutes les questions/patterns ==> input_texts.
input_texts=[]
for intent in data['intents']:
  for question in intent['patterns']:
    input_texts.append(question)
print(input_texts)
print(len(input_texts))
```

```python
#Une List qui contient toutes les intentions (uniques) ==> classes
classes=[]
for intent in data['intents']:
  classes.append(intent['tag'])
print(classes)
```

```python
#Une List qui contient l’intention correspondante pour chaque questions/patterns
output_intentions=[]
for intent in data['intents']:
  for question in intent['patterns']:

    output_intentions.append(intent['tag'])

print(output_intentions)
print(len(output_intentions))
```

```python
#Appliquer la tokenization par mot et la lemmatisation francais
import spacy

# Charger le modèle français
nlp = spacy.load("fr_core_news_sm")

list_tockens=[]
for intent in data['intents']:
  for question in intent['patterns']:
    # Analyser le texte
    doc = nlp(question.lower())
    for token in doc:
      list_tockens.append(token.lemma_)

print(list_tockens)
print(f"size tokens : {len(list_tockens)}")
```

```python
#Supprimer les tokens qui contient seulement les symboles.
import string
list_tockens = [t for t in list_tockens if t not in string.punctuation]
print(list_tockens)
```

```python
print(len(list_tockens))
```

```python
# laisser seulement les valeurs uniques dans la liste list_tockens
list_tockens = list(dict.fromkeys(list_tockens))

print(list_tockens)
```

```python
print(len(list_tockens))
```

```python
'''
Une List training qui contient les représentations numériques de chaque
question/pattern X et les représentations numérique des intentions
correspondantes Target. Exemple :
Training = [ [ [X1] , [Target1] ],
 [ [X2] , [Target2] ],
 [ [X3] , [Target3] ],
 [ [X4] , [Target4] ],
 [ [X5] , [Target5] ],
 ... ]
'''
training = []
for phrase in input_texts :
  vect_x=[0 for i in range(len(list_tockens))]
  tok = nlp(phrase.lower())

  vect_y=[0 for i in range(len(classes))]
  idx=input_texts.index(phrase)
  intent=output_intentions[idx]
  idx=classes.index(intent )
  vect_y[idx]=1
  #print(vect_y)

 # print(t)
  for item in tok:
        result=item.lemma_
        if result not in string.punctuation:
            if result in list_tockens:
                idx=list_tockens.index(result)
                #print(idx)
                vect_x[idx]=1
  training.append([vect_x, vect_y])

  #print(vect_x)
  #print(len(input_texts))
print(training)
print(len(training))




```

```python
'''
Mélanger les éléments de la List Training en utilisant :
random.shuffle(training)
'''
import random
import numpy as np

print(len(training))
# mélanger les données et les convertir en array
random.shuffle(training)
'''
Convertir la list Training en Array.

'''
training = np.array(training, dtype=object)
print(training)
```

```python
# séparer les features et les labels target
train_X = np.array(list(training[:, 0]))
train_y = np.array(list(training[:, 1]))
```

```python
input_size = (len(train_X[0]),)
output_shape = len(train_y[0])
epochs = 10
```

```python
import tensorflow as tf
from keras.models import Sequential
from keras.layers import Dense, Embedding, LSTM
```

```python
model = Sequential()
model.add(Dense(128, input_shape=input_size, activation="relu"))
model.add(Dense(64,  activation="relu"))
model.add(Dense(output_shape,  activation="softmax"))

ada = tf.keras.optimizers.Adam(learning_rate=0.01)
model.compile(loss='crossentropy', optimizer=ada, metrics=['accuracy'])
```

```python
print(model.summary())
```

```python
model.fit(x=train_X, y=train_y, epochs=epochs, verbose=1 )
```
